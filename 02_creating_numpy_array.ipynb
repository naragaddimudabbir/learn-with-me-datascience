## 🧱 Creating NumPy Arrays – Simplified Guide

### 1. **How to Create NumPy Arrays**

You can create arrays by converting Python lists or using NumPy’s built-in functions:

* `np.zeros()` → Array filled with 0s
* `np.ones()` → Array filled with 1s
* `np.full()` → Array filled with a custom value
* `np.eye()` → Identity matrix (used in linear algebra)
* `np.arange()` → Array with a range of values
* `np.linspace()` → Array with evenly spaced values

These methods save time and avoid manual loops.

---

### 2. **Checking Array Properties**

Every NumPy array has useful info:

* `shape` → Rows and columns
* `size` → Total number of elements
* `ndim` → Number of dimensions (1D, 2D, etc.)
* `dtype` → Data type of elements

Knowing these helps you write better and faster code.

---

### 3. **Changing Data Types**

You can set or convert array data types like `int32`, `float64`, etc.
This helps control memory use and precision, especially with large datasets.

---

### 4. **Reshaping & Flattening**

* **Reshape**: Change the structure (e.g., 1D to 2D)
* **Flatten**: Turn multi-dimensional arrays into a 1D array

Useful in machine learning and data preprocessing.

---

### ✅ Key Takeaways

* NumPy arrays are faster than lists because they use fixed types.
* Built-in creation methods make code cleaner and quicker.
* Understanding shapes and types boosts performance.
* Reshape and flatten help manage data structure easily.
