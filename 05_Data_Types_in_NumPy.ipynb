## 🧠 NumPy Data Types (Simplified)

### 1. 🧾 What Are NumPy Data Types?

* NumPy arrays are **homogeneous** – all elements must be of the **same type**.
* Python lists can have mixed types, but NumPy arrays **cannot**.

### 🧩 Common Data Types

| Type                      | Use Case                              |
| ------------------------- | ------------------------------------- |
| `int32`, `int64`          | Whole numbers (different sizes)       |
| `float32`, `float64`      | Decimal numbers (different precision) |
| `bool`                    | True/False values                     |
| `complex64`, `complex128` | Complex numbers like 1 + 2j           |
| `object`                  | Mixed types (slowest)                 |
| `str` or `'U'`            | Strings (fixed-length)                |

🧪 **Example**:

```python
import numpy as np
arr = np.array([1, 2, 3])
print(arr.dtype)  # int64 or int32 depending on your system
```

---

### 2. 🔄 Changing Data Types

Use `.astype()` to **convert** data types (called casting).

🧪 **Example: Float to Int**

```python
arr = np.array([1.5, 2.7, 3.9])
arr_int = arr.astype(np.int32)
print(arr_int)  # [1 2 3]
```

🧪 **Example: Reduce Memory (Downcasting)**

```python
arr = np.array([1000000, 2000000], dtype=np.int64)
arr_small = arr.astype(np.int32)
print(arr_small.dtype)  # int32
```

---

### 3. 🧠 Why Data Types Matter

| Factor       | Impact                             |
| ------------ | ---------------------------------- |
| Memory usage | Smaller types use **less memory**  |
| Speed        | Smaller types are **faster**       |
| Precision    | Bigger types are **more accurate** |

🧪 **Example: Memory Comparison**

```python
a = np.array([1, 2, 3], dtype=np.int64)  # 24 bytes
b = np.array([1, 2, 3], dtype=np.int32)  # 12 bytes
```

---

### 4. 🧵 Strings in NumPy

```python
arr = np.array(['apple', 'banana'], dtype='U10')
print(arr)
```

* Fixed-length strings. `'U10'` means up to 10 characters.
* Not ideal for text-heavy tasks. Use lists or pandas for better performance.

---

### 5. ⚡ Complex Numbers

```python
arr = np.array([1+2j, 3+4j], dtype='complex128')
print(arr)
```

* Useful for scientific computing involving imaginary numbers.

---

### 6. 🎒 Object Data Type

```python
arr = np.array([{'a': 1}, [1, 2], 'hi'], dtype=object)
print(arr)
```

* Supports **mixed** types.
* Slower, use **only when necessary**.

---

### 7. ✅ Choosing the Right Type

* **Need speed & small size?** Use `int32`, `float32`.
* **Need precision?** Use `float64`, `complex128`.
* **Need mixed content?** Use `dtype=object`, but carefully.

---

### 🧭 Summary

* NumPy arrays must use a **single data type**.
* Use `.astype()` to **change types** when needed.
* Right data types = **faster + smaller + accurate**.
